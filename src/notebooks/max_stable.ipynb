Pseudo ortogonalne

```python
for seed in seeds:
    images = create_pseudo_orthogonal_images(seed)
    for i in range(max_number_of_patters):
        is_stable, disp = train_hopfield_from_array_for_stability(images[0:i + 1, :])
        if is_stable == False:
            print(f"seed = {seed}")
            print(f"Stopped at: {i} | {disp}")
            print("--------------------------------------------")
            break

#plot_patterns(images, (5,5))
```
