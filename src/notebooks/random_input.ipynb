```python
import os
import sys
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from model.hopfield_network import HopfieldNetwork
from notebooks.plot_patterns import plot_patterns
from notebooks.helpers import create_number_of_random_images, read_data_train_hebb, \
    classify_recalls
```

```python
seed = 1
how_many = 100
```

Large 25x25

```python
name = "large-25x25.csv"
n = 25 * 25
images = create_number_of_random_images(n, seed, how_many)

hn, patterns = read_data_train_hebb(dataset_name=name, is_async=True)

results = np.ndarray((how_many, n))
for i in range(how_many):
    result = hn.recall(images[i], 100)
    # display(f"Result: {result["Result"]}. Energy value: {result["Energy"]}")
    results[i, :] = result["History"][-1]

classify_recalls(patterns, results, (25, 25));
```
