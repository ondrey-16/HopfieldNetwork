```python
import os
import sys
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from model.hopfield_network import HopfieldNetwork
from notebooks.plot_patterns import plot_patterns
```

```python
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))

if project_root not in sys.path:
    sys.path.append(project_root)
```

```python
patterns_path = Path(project_root) / "data/small-7x7.csv"
```

```python
df = pd.read_csv(patterns_path, header=None)
df.head(20)
```

```python
patterns = np.asarray(df, dtype=float)
plot_patterns(patterns, (7, 7))
```

```python
hn = HopfieldNetwork(n=49, method="hebb", if_asynchronic=False, random_state=42)
```

```python
hn.train(patterns=patterns)
```

```python
noise_pattern = patterns[2]

rng = np.random.default_rng(123)

noise_level = 0.1
noisy_pattern = noise_pattern.copy()

n_flips = round(noise_level * noisy_pattern.size)

indices = rng.choice(
    noisy_pattern.size,
    size=n_flips,
    replace=False
)

noisy_pattern[indices] *= -1
```

```python
result = hn.recall(noisy_pattern, 100)
hn.plot_history(result["History"], (7, 7))
```
